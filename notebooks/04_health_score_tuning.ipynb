# Health Score Weight Calibration

Tuning the composite health score weights for churn prediction accuracy.

```python
from analytics.health_scoring import compute_health_from_parquet
import plotly.express as px

scores = compute_health_from_parquet(
    'data/users.parquet',
    'data/events.parquet',
    'data/workspaces.parquet'
)
print(scores['risk_tier'].value_counts())

fig = px.histogram(scores, x='health_score', color='risk_tier',
                   title='Health Score Distribution by Risk Tier')
fig.show()
```

```python
# Top at-risk workspaces
at_risk = scores[scores['risk_tier'].isin(['Critical', 'At Risk'])]
print(f'At-risk workspaces: {len(at_risk):,} ({len(at_risk)/len(scores)*100:.1f}%)')
at_risk.head(20)
```
