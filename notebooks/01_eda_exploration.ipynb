# ConnectHub EDA — Initial Data Exploration

Exploring the synthetic dataset: user distributions, event patterns, feature usage.

```python
import pandas as pd
import plotly.express as px

users = pd.read_parquet('data/users.parquet')
events = pd.read_parquet('data/events.parquet')

print(f'Users: {len(users):,}')
print(f'Events: {len(events):,}')
print('\nPlan distribution:')
print(users['plan_tier'].value_counts())
```

```python
# Event distribution
top_events = events['event_name'].value_counts().head(15)
fig = px.bar(x=top_events.index, y=top_events.values, title='Top 15 Event Types')
fig.show()
```

```python
# Daily event volume
daily = events.groupby('event_date').size().reset_index(name='count')
fig = px.line(daily, x='event_date', y='count', title='Daily Event Volume')
fig.show()
```
